## 18. Función set()

**¿Qué es?**

Es una función incorporada en Python que permite crear un conjunto de elementos únicos, eliminando los valores repetidos

```python
animales = ["Perro", "Gato", "Perro", "Conejo", "Gato"]

animales_unicos = set(animales)

print("Animales sin repetir:")
print(animales_unicos)
```

## 20. Función sorted()

**¿Qué es?**

Es una función incorporada en Python que permite ordenar los elementos de una colección y generar una nueva lista ordenada

**Ejemplo práctico:**

## 21. Función reversed()

**¿Qué es?**

Es una función incorporada en Python que permite obtener los elementos de una secuencia en orden inverso

**Uso:**

Se utiliza para recorrer listas, tuplas y otras secuencias desde el último elemento hasta el primero, sin modificar la colección original

**Ejemplo práctico:**

Permite mostrar una lista de animales comenzando por el último elemento y terminando con el primero

```python
animales = ["Perro", "Gato", "Caballo", "Conejo"]

invertidos = list(reversed(animales))

print("Lista en orden inverso:")
print(invertidos)
```

## 22. Función enumerate()

**¿Qué es?**

Es una función incorporada en Python que permite recorrer una colección de elementos obteniendo su posición y su valor al mismo tiempo

**Uso:**

Se utiliza para enumerar los elementos de una lista, facilitando su identificación mediante un número de posición

**Ejemplo práctico:**

Permite enumerar una lista de animales para organizar y presentar cada uno con su respectivo número

```python
animales = ["Perro", "Gato", "Caballo", "Conejo"]

for numero, animal in enumerate(animales, start=1):
    print(numero, animal)
```

## 23. Función zip()

**¿Qué es?**

Es una función incorporada en Python que permite combinar dos o más colecciones de elementos en pares o grupos, según sus posiciones

**Uso:**

Se utiliza para relacionar información que se encuentra distribuida en diferentes listas o secuencias

**Ejemplo práctico:**

Permite relacionar los nombres de varios animales con sus respectivas edades, facilitando la organización de la información

```python
animales = ["Perro", "Gato", "Caballo"]
edades = [3, 2, 5]

resultado = list(zip(animales, edades))

print("Animales y sus edades:")
print(resultado)
```

## 24. Función any()

**¿Qué es?**

Es una función incorporada en Python que permite comprobar si al menos uno de los elementos de una colección tiene un valor verdadero

**Uso:**

Se utiliza para verificar si existe al menos un elemento que cumpla una condición dentro de una lista u otra colección

**Ejemplo práctico:**

Permite comprobar si al menos un animal de una lista tiene una edad mayor a cinco años

```python
edades = [2, 3, 7, 4]

resultado = any(edad > 5 for edad in edades)

print("¿Hay algún animal mayor de 5 años?", resultado)
```

## 25. Función all()

**¿Qué es?**

Es una función incorporada en Python que permite comprobar si todos los elementos de una colección tienen un valor verdadero

**Uso:**

Se utiliza para verificar si todos los elementos de una lista cumplen una condición determinada

**Ejemplo práctico:**

Permite comprobar si todos los animales de una lista tienen una edad mayor que cero

```python
edades = [2, 3, 7, 4]

resultado = all(edad > 0 for edad in edades)

print("¿Todos los animales tienen más de 0 años?", resultado)
```

## 26. Función filter()

**¿Qué es?**

Es una función incorporada en Python que permite filtrar los elementos de una colección según una condición específica

**Uso:**

Se utiliza para seleccionar únicamente los elementos que cumplen un criterio determinado, descartando los que no lo cumplen

**Ejemplo práctico:**

Permite seleccionar los animales que tienen una edad mayor a tres años dentro de una lista

```python
edades = [2, 5, 3, 7, 1, 6]

mayores = list(filter(lambda edad: edad > 3, edades))

print("Edades mayores a 3 años:")
print(mayores)
```

## 27. Función divmod()

**¿Qué es?**

Es una función incorporada en Python que permite realizar una división y obtener dos resultados: el cociente y el residuo

**Uso:**

Se utiliza para conocer cuántas veces un número cabe dentro de otro y cuánto sobra después de realizar la división

**Ejemplo práctico:**

Permite calcular cuántos grupos completos se pueden formar con una cantidad determinada de animales y cuántos animales quedan sin agrupar

```python
animales = 17
grupos = 5

resultado = divmod(animales, grupos)

print("Grupos completos y animales restantes:")
print(resultado)
```

## 28. Función isinstance()

**¿Qué es?**

Es una función incorporada en Python que permite comprobar si un objeto pertenece a un tipo de dato determinado

**Uso:**

Se utiliza para verificar si una variable corresponde a un número entero, decimal, texto, lista u otro tipo de dato

```python
edad = 8
nombre = "coco"

print(isinstance(edad, int))
print(isinstance(nombre, str))
print(isinstance(edad, str))
```

## 30. Función map()

**¿Qué es?**

Es una función incorporada en Python que permite aplicar una operación o función a cada elemento de una colección

**Uso:**

Se utiliza para transformar los elementos de una lista sin tener que escribir un ciclo de repetición manualmente

**Ejemplo práctico:**

Permite convertir las edades de varios animales de años a meses, multiplicando cada valor por doce

```python
edades = [1, 2, 3, 4]

edades_meses = list(map(lambda edad: edad * 12, edades))

print("Edades de los animales en meses:")
print(edades_meses)
```
